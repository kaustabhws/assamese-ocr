# 02 — Train the Assamese recognizer

This trains the compact CRNN-CTC recognizer. It selects checkpoints by validation CER and does not inspect test metrics.

```python
import torch

assert torch.cuda.is_available(), "Select a GPU runtime before training"
print(torch.cuda.get_device_name(0))
```

```python
from axomiya_ocr.data.vocab import Vocabulary
from axomiya_ocr.training.config import load_config
from axomiya_ocr.training.engine import build_model

config = load_config("configs/recognizer.yaml")
vocab = Vocabulary.load(config["data"]["vocab_path"])
model = build_model(config, vocab)
print(f"classes={vocab.size}, parameters={model.parameter_count():,}")
assert "ৰ" in vocab.characters and "ৱ" in vocab.characters
```

```python
from axomiya_ocr.training.engine import train

result = train(config)
result
```

```python
import json
from pathlib import Path

import pandas as pd

history = json.loads(Path("artifacts/recognizer/history.json").read_text(encoding="utf-8"))
pd.DataFrame(history).plot(x="epoch", y=["train_loss", "cer"], secondary_y="cer", grid=True)
```
